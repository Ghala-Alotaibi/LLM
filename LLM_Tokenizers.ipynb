```python
from transformers import AutoTokenizer
```

```python
sentence = "Hello world"
```

```python
# load the pretrained tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")
```

```python
# apply the tokenizer to the sentence and extract the token ids
token_ids = tokenizer(sentence).input_ids
```

```python
print(token_ids)
```

```python
#To map each token ID to its corresponding token we use decode method of the tokenizer.so it looks in the same vocabulary in reverse:
for id in token_ids:
    print(tokenizer.decode(id)) #sep = end of sentence
```

## Visualizing Tokenization, to visually see where one token ends and the next begins.

```python
# A list of colors in RGB for representing the tokens
colors = [
    '102;194;165', '252;141;98', '141;160;203',
    '231;138;195', '166;216;84', '255;217;47'
]

def show_tokens(sentence: str, tokenizer_name: str):
    """ Show the tokens each separated by a different color """

    # Load the tokenizer and tokenize the input
    tokenizer = AutoTokenizer.from_pretrained(tokenizer_name)
    token_ids = tokenizer(sentence).input_ids

    # Extract vocabulary length
    print(f"Vocab length: {len(tokenizer)}")

    # Print a colored list of tokens
    for idx, t in enumerate(token_ids):
        print(
            f'\x1b[0;30;48;2;{colors[idx % len(colors)]}m' +
            tokenizer.decode(t) +
            '\x1b[0m',
            end=' '
        )
```

```python
text = """
English and CAPITALIZATION
🎵 鸟
show_tokens False None elif == >= else: two tabs:"    " Three tabs: "       "
12.0*50=600
"""
```

##Explore tokenizing models


```python
show_tokens(text, "bert-base-cased")
```

```python
show_tokens(text, "bert-base-uncased")
```

```python
show_tokens(text, "Xenova/gpt-4")
```

```python
show_tokens(text, "gpt2")
```

```python
show_tokens(text, "google/flan-t5-small")
```
